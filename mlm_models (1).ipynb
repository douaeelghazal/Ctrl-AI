# Masked Language Modeling (MLM)

This notebook demonstrates masked language modeling techniques.
Masked language modeling predicts a masked token in a sequence, and the model can attend to tokens bidirectionally.

```python
from datasets import load_dataset
from huggingface_hub import login

# Use your Hugging Face token here
# login(token="YOUR_HUGGINGFACE_TOKEN")

# Load dataset
dataset = load_dataset('wikitext', 'wikitext-2')
```

```python
from transformers import AutoTokenizer, AutoModelForMaskedLM

# Load pre-trained model and tokenizer
model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForMaskedLM.from_pretrained(model_name)
```

```python
# Example: Mask tokens and predict
text = "The quick brown fox jumps over the lazy dog"
tokens = tokenizer.encode(text, return_tensors='pt')
# Mask a token
masked_tokens = tokens.clone()
masked_tokens[0, 3] = tokenizer.mask_token_id
print(f"Original: {text}")
print(f"Masked input: {tokenizer.decode(masked_tokens[0])}")
```
